```python
def top_categories(data, col, n=10):
    return data[col].value_counts().head(n).index
```

```python
def median_order(data, cat_col, target_col):
    return (data.groupby(cat_col)[target_col].median().sort_values().index)
```

BLOCK 1 Обзор рынка

График 1. Какой диапазон цены за квадратный метр встречается чаще всего?

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['price_per_m2'], bins=40, kde=True)
plt.title('Distribution of Price per m²')
plt.xlabel('Price per m²')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

Most apartments are concentrated around the middle price-per-square-meter range, with a visible peak roughly between 7 and 10 euro per m². The distribution is strongly right-skewed, which suggests that premium listings are relatively rare but still significantly extend the upper tail of the market. Because of this skewness, median values may be more informative than means in further comparisons.

price_per_m2 is used as the main pricing metric because it allows fairer comparison between apartments of different sizes.

График 2. Какие площади квартир встречаются чаще всего?

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['living_space'], bins=40, kde=True)
plt.title('Distribution of Living Space')
plt.xlabel('Living Space, m²')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

Most apartments in the dataset are concentrated around the medium living space range, with the highest density roughly between 50 and 70 m². The distribution is right-skewed, indicating that large apartments are relatively rare compared to smaller and medium-sized properties.

BlOCK 2 География цен

График 3. Насколько сильно отличается цена за квадратный метр между крупными городами?

```python
top_cities = top_categories(df, 'city', 10)
city_df = df[df['city'].isin(top_cities)].copy()

city_order = (
    city_df.groupby('city')['price_per_m2']
    .median()
    .sort_values(ascending=False)
    .index
)

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=city_df,
    x='city',
    y='price_per_m2',
    order=city_order,
    showfliers=False
)
plt.title('Price per m² by City')
plt.xlabel('City')
plt.ylabel('Price per m²')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

There are clear geographic differences in rental prices across cities. München has the highest median price_per_m2, followed by Frankfurt am Main, while eastern cities such as Chemnitz, Magdeburg, and Halle Saale remain much more affordable. This suggests that location is one of the strongest drivers of rental pricing in the dataset.

График 4. Как различается медианная цена аренды за квадратный метр между федеральными землями?

```python
state_price = (
    df.groupby('state')
    .agg(
        median_price_per_m2=('price_per_m2', 'median'),
        listings=('state', 'size')
    )
    .sort_values('median_price_per_m2', ascending=False)
    .reset_index()
)

plt.figure(figsize=(12, 6))
sns.barplot(
    data=state_price,
    x='state',
    y='median_price_per_m2'
)
plt.title('Median Price per m² by State')
plt.xlabel('State')
plt.ylabel('Median Price per m²')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The bar chart shows substantial variation in median price_per_m2 across federal states. Hamburg and Berlin rank highest, while Thüringen, Sachsen-Anhalt, and Sachsen appear at the lower end of the distribution. The gap between the most and least expensive states is large, indicating that regional location is a major determinant of rental market pricing in the dataset.

Блок 3. Площадь, комнаты и общая аренда

График 5. Как меняется общая аренда при увеличении площади квартиры?

```python
plt.figure(figsize=(10, 6))
plt.hexbin(
    df['living_space'],
    df['total_rent'],
    gridsize=40,
    mincnt=1
)
plt.colorbar(label='Count')
plt.title('Living Space vs Total Rent')
plt.xlabel('Living Space, m²')
plt.ylabel('Total Rent')
plt.tight_layout()
plt.show()
```

Living_space is strongly associated with total_rent, confirming that apartment size is one of the main drivers of overall rental cost. At the same time, the spread of rent values increases for larger properties, which suggests that other factors such as location, quality, and property type also play an important role in pricing.

График 6. Как площадь связана с ценой за квадратный метр?

```python
plt.figure(figsize=(10, 6))
plt.hexbin(
    df['living_space'],
    df['price_per_m2'],
    gridsize=40,
    mincnt=1
)
plt.colorbar(label='Count')
plt.title('Living Space vs Price per m²')
plt.xlabel('Living Space, m²')
plt.ylabel('Price per m²')
plt.tight_layout()
plt.show()
```

График 7. Как меняется цена за квадратный метр в зависимости от числа комнат?

```python
room_order = (df.groupby('no_rooms')['price_per_m2'].median().sort_index().index)

plt.figure(figsize=(10, 6))
sns.boxplot(
    data=df,
    x='no_rooms',
    y='price_per_m2',
    order=room_order,
    showfliers=False
)
plt.title('Price per m² by Number of Rooms')
plt.xlabel('Number of Rooms')
plt.ylabel('Price per m²')
plt.tight_layout()
plt.show()
```

The number of rooms is not strictly discrete in this dataset due to local real estate conventions. Rather than forcing rounding during cleaning, the original values were preserved and grouped into broader categories for clearer analysis.

График 8.

```python
def room_group(x):
    if x < 2:
        return '1'
    elif x < 3:
        return '2'
    elif x < 4:
        return '3'
    elif x < 5:
        return '4'
    else:
        return '5+'

df['rooms_group'] = df['no_rooms'].apply(room_group)
```

```python
sns.boxplot(
    data=df,
    x='rooms_group',
    y='price_per_m2',
    order=['1', '2', '3', '4', '5+'],
    showfliers=False
)
plt.title('Price per m² by Number of Rooms')
plt.xlabel('Number of Rooms')
plt.ylabel('Price per m²')
plt.tight_layout()
plt.show()
```

One-room apartments have the highest median price per square meter and also show the greatest price variability. In contrast, apartments with two or more rooms are priced more similarly,which suggests that the main unit-price premium is concentrated in the most compact housing segment.

BlOCK 4 Качество жилья

График 9. Связаны ли качество, состояние и новизна жилья с более высокой ценой за квадратный метр?

```python
condition_order = median_order(df.dropna(subset=['condition']), 'condition', 'price_per_m2')
interior_order = median_order(df.dropna(subset=['interior_qual']), 'interior_qual', 'price_per_m2')

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

sns.boxplot(
    data=df,
    x='condition',
    y='price_per_m2',
    order=condition_order,
    showfliers=False,
    ax=axes[0]
)
axes[0].set_title('Price per m² by Condition')
axes[0].set_xlabel('Condition')
axes[0].set_ylabel('Price per m²')
axes[0].tick_params(axis='x', rotation=45)

sns.boxplot(
    data=df,
    x='interior_qual',
    y='price_per_m2',
    order=interior_order,
    showfliers=False,
    ax=axes[1]
)
axes[1].set_title('Price per m² by Interior Quality')
axes[1].set_xlabel('Interior Quality')
axes[1].set_ylabel('Price per m²')
axes[1].tick_params(axis='x', rotation=45)

sns.boxplot(
    data=df,
    x='newly_const',
    y='price_per_m2',
    showfliers=False,
    ax=axes[2]
)
axes[2].set_title('Price per m² by Newly Constructed')
axes[2].set_xlabel('Newly Constructed')
axes[2].set_ylabel('Price per m²')

plt.tight_layout()
plt.show()
```

The three boxplots consistently show that higher-quality housing attributes are associated with higher rental prices per square meter. Better overall condition, higher interior quality, and newly constructed status all correspond to a clear pricing premium. This suggests that beyond location and apartment size, property quality is one of the key determinants of rental value in the dataset.

BLOCK 5 Возраст недвижимости

График 10. Как меняется медианная цена за квадратный метр в зависимости от возраста дома?

```python
age_df = df[df['property_age'].notna()].copy()

age_bins = [0, 5, 10, 20, 30, 50, 80, 120, 200]
age_labels = ['0-5', '6-10', '11-20', '21-30', '31-50', '51-80', '81-120', '121+']

age_df['age_group'] = pd.cut(
    age_df['property_age'],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

age_summary = (
    age_df.groupby('age_group', observed=False)['price_per_m2']
    .median()
    .reset_index()
)

plt.figure(figsize=(10, 6))
sns.lineplot(
    data=age_summary,
    x='age_group',
    y='price_per_m2',
    marker='o'
)
plt.title('Median Price per m² by Property Age Group')
plt.xlabel('Property Age Group')
plt.ylabel('Median Price per m²')
plt.tight_layout()
plt.show()
```

Newer properties have higher median prices per square meter, while older buildings tend to be cheaper. The strongest decline appears after the 11–20 year group, after which prices stabilize at a lower level.

BLOCK 6 Общая числовая картина

График 11.Какие числовые признаки сильнее всего связаны друг с другом, особенно с ценой и арендой?

```python
num_cols = [
    'service_charge',
    'total_rent',
    'base_rent',
    'living_space',
    'no_rooms',
    'floor',
    'number_of_floors',
    'price_per_m2',
    'cold_rent_per_m2',
    'property_age',
    'room_density'
]

corr = df[num_cols].corr(numeric_only=True)

plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', square=True)
plt.title('Correlation Matrix of Numerical Features')
plt.tight_layout()
plt.show()
```

The correlation matrix confirms several key patterns in the dataset. Total rent is strongly associated with apartment size and base rent, while price per square meter is much less dependent on living space itself. Property age has a moderate negative relationship with unit price, supporting the earlier finding that newer apartments tend to command a premium. The apartment's floor level does not in itself appear to be a major factor influencing its value. Overall, the heatmap suggests that apartment size is the main driver of total rent, while unit price is more strongly shaped by location and quality-related factors than by size alone.
